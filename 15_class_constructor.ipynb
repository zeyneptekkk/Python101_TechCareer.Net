```python
class ArabaConstructor:
    
    def __init__(self, marka="bilinmiyor",model="bilinmiyor", yil=0, renk="bilinmiyor"):  
        self.marka=marka
        self.model=model
        self.yil=yil
        self.renk=renk
        print("Araba sınıfına yeni bir araba oluşturuldu")
        
    def bilgileri_goster(self):
        print(f"Marka: {self.marka} \nModel: {self.yil} \nRenk: {self.renk} \nYıl: {self.yil}")
             
    def set_marka(self,marka):
        self.marka=marka
        
    def get_marka(self,marka):
        return self.marka      
    
                 
    def set_model(self,model):
        self.model=model
        
    def get_model(self,model):
        return self.model      
                 
        
    def set_renk(self,renk):
        self.renk=renk
        
    def get_renk(self,renk):
        return self.renk
                 
        
    def set_yil(self,yil):
        if yil >= 2020:
            self.yil=yil
        else:
            print("2020 nin altındaki arabalar alınmıyor")
        self.yil=yil
        
    def get_yil(self,yil):
        return self.yil     
    
    
    
```

```python
araba1=ArabaConstructor()
araba1.bilgileri_goster()
```

```python
araba2=ArabaConstructor()
araba2.set_yil(2000)
araba2.set_model("BMW")
araba2.set_renk("Gri")
araba2.set_marka(2000)

araba2.bilgileri_goster()
```
